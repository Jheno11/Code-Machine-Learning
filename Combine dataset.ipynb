```python
import pandas as pd
import numpy as np

deg = pd.read_csv('degs.csv', usecols=['Ensembl_ID'])
dataset = pd.read_csv('datasetWhiteMerged.csv')

counts = deg.set_index('Ensembl_ID')

count = dataset.set_index('Ensembl_ID')

combined = pd.merge(counts, count, left_index=True, right_index=True, how='inner')
combined
```

```python
combined.to_csv('Data139.csv')
```
