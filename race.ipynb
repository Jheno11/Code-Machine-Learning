```python
import pandas as pd

phenotype = pd.read_csv("phenotype.csv")

gene = pd.read_excel("dataset.xlsx")

gene_data = gene

white_samples = phenotype[phenotype["race.demographic"].str.lower() == "white"]["sample"]

common_samples = set(white_samples) & set(gene_data.columns)

gene_white = gene_data[["Ensembl_ID"] + list(common_samples)]

gene_white.to_excel("white_races.xlsx", index=False)
```
