```python
import pandas as pd

phenotype = pd.read_csv("phenotype.csv")

mirna = pd.read_excel("miRNA dataset.xlsx", sheet_name="TCGA-PRAD.mirna")

mirna_data = mirna

white_samples = phenotype[phenotype["race.demographic"].str.lower() == "white"]["sample"]

common_samples = set(white_samples) & set(mirna_data.columns)

mirna_white = mirna_data[["miRNA_ID"] + list(common_samples)]

mirna_white.to_excel("miRNA_white_races.xlsx", index=False)
```
